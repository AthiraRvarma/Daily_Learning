```python
#two dimensional array
b=np.array([[1,2,3],[4,5,6]])
print(b)
print(b.ndim)#print dimesion of array
print(b.shape)#print shape of array means row by column
print(b.size)#print totsl no of elements in an array
print(b.itemsize)#print single element size in byte
print(b.nbytes)#print total element size in bytes.
print(b.T)#print transpose of matrix.
print(b.T.shape)
#for matrix multiplication we use @ symbol
```

```python
#array creating functions
a=np.zeros(10)
print(a)
b=np.ones((2,3))#create a 2row 3 column array with having all element 1
print(b)
c=np.full((3,2),3.14)
print(c)
#creatng identity matrics we use 'eye'keyword
d=np.eye(3)
print(d)
```

```python
#accessing array elements.
a=np.arange(10)
print(a[1])
print(a[-1])
print(a[-2])
print(a[-5])
#slicing
b=a[1:5]#slicing for this give starting index and ending index=n-1
print(b)
```

```python
#reshape function
a=np.arange(25).reshape(5,5)
print(a)
#2D array slicing.
print(a[2:4,4:])
print(a[1:4:2,1::3])
print(a[1::3,0::4])
```

```python
#Pandas
import pandas as pd
import numpy as np
#create empty series.
ser=pd.Series()
print(ser)
data=np.array(["a","b","c","d"])
new_ser=pd.Series(data)
print(new_ser)
#create a series from dictionary
info={"name":"Rudra","age":3,"class":"Prekg"}
a=pd.Series(info)
print(a)
#create series using tuple
l=pd.Series((3,4))
print(l)
#create series using scalar
l1=pd.Series(2,index=[0,1,2,3,4])
print(l1)
l2=pd.Series([4,5,6],index=["r","g","b"])
print(l2)
#create a data frame.
list=["apple","orenge","banana"]
fruits=pd.Series(list)
print(fruits)
data_frame=pd.DataFrame(list)#in case of data frame column index is present in the output.
print(data_frame)
Student_info={"Id":[101,102,103],
              "name":["Shana","Arya","Hanan"],
              "Department":["bsc","pgd","cse"]}
Student_table=pd.DataFrame(Student_info)
print(Student_table)
#accessing element from Data frame.
print(Student_table.name[0])
print(Student_table.Department[2])
info={
    "fruits":pd.Series(["apple","orenge","banana"]),
    "vegetables":pd.Series(["brinjal","beens","drumstick"])
    }
df=pd.DataFrame(info)
print(df)
#adding new column to existing data frame object
print("Add new column by passing series")
df["millets"]=pd.Series(["rice","wheat","maize"])
print(df)
```

```python

nums={
    "one":pd.Series([1,2,3],index=["a","b","c"]),
    "two":pd.Series([4,5,6],index=["a","b","c"])
}
df1=pd.DataFrame(nums)
print(df1)
df1["Three"]=df1["one"]+df1["two"]
print(df1)
```

```python
#using del function.
print("delete the first column: ")
del df1["one"]
print(df1)
```
