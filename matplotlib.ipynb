```python
#matplotlib
import matplotlib.pyplot as plt
%matplotlib inline
y=(2,4,3,6,8,3,5,9)
plt.plot(y)
```

```python
x=(0,1,2,3,4,5,6,7)
y=(2,4,3,6,8,3,5,9)
plt.plot(x,y)
```

```python
import numpy as np
x=np.linspace(0,3*np.pi,50)
y=np.sin(x)
plt.plot(x,y)
```

```python
x=np.linspace(0,10,80)
y=np.sin(x)
z=np.cos(x)
plt.plot(x,y, 'r--')
plt.plot(x,z, 'yo')#yo=>yello dot
plt.xlabel("x axis label")
plt.ylabel("y axis label")
plt.title("sin and cos")
plt.legend(["sin","cos"])
plt.savefig("sincos.png")
```

```python
x=np.linspace(0,10,20)
plt.plot(x,x,'r--')
plt.plot(x,x**2,'bs')#bs=>blue square
plt.plot(x,x**3,'g^')#g^=>green triangle
```

```python
x=np.linspace(0,10,80)
y=np.sin(x)
z=np.cos(x)
plt.plot(x,y, 'r--')
plt.xlabel("x axis label")
plt.ylabel("y axis label")
plt.title("sin")
plt.legend(["sin"])
plt.figure()
plt.plot(x,z, 'yo')#yo=>yello dot
plt.xlabel("x axis label")
plt.ylabel("y axis label")
plt.title("cos")
plt.legend(["cos"])
plt.savefig("cos.png")
```

```python
plt.subplot(1,2,1)
plt.plot(x,y, 'r--')
plt.xlabel("x axis label")
plt.ylabel("y axis label")
plt.title("sin")
plt.legend(["sin"])
plt.subplot(1,2,2)
plt.plot(x,z, 'yo')#yo=>yello dot
plt.xlabel("x axis label")
plt.ylabel("y axis label")
plt.title("cos")
plt.legend(["cos"])
plt.savefig("cos.png")
```

```python
plt.subplot(2,1,1)
plt.plot(x,y, 'r--')
plt.xlabel("x axis label")
plt.ylabel("y axis label")
plt.title("sin")
plt.legend(["sin"])
plt.subplot(2,1,2)
plt.plot(x,z, 'yo')#yo=>yello dot
plt.xlabel("x axis label")
plt.ylabel("y axis label")
plt.title("cos")
plt.legend(["cos"])
plt.savefig("cos.png")
```
